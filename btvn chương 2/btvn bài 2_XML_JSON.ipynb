## Bài 2.3 – 2.5 – XML + minidom

```python
# Tạo file sample.xml
xml_content = """<?xml version="1.0"?>
<company>
    <name>GeeksForGeeks Company</name>
    <staff id="1">
        <name>Amar Pandey</name>
        <salary>8.5 LPA</salary>
    </staff>
    <staff id="2">
        <name>Akbar Khan</name>
        <salary>6.5 LPA</salary>
    </staff>
    <staff id="3">
        <name>Anthony Walter</name>
        <salary>3.2 LPA</salary>
    </staff>
</company>
"""

with open("sample.xml", "w", encoding="utf-8") as f:
    f.write(xml_content)

print("Đã tạo file sample.xml")
```

```python
import xml.dom.minidom

doc = xml.dom.minidom.parse("sample.xml")

print("Root:", doc.documentElement.tagName)

staffs = doc.getElementsByTagName("staff")
for s in staffs:
    name = s.getElementsByTagName("name")[0].firstChild.nodeValue
    salary = s.getElementsByTagName("salary")[0].firstChild.nodeValue
    print(name, "-", salary)
```

## Bài 2.6 – Đọc RSS Feed

```python
import feedparser

url = "http://hindustantimes.com/rss/topnews/rssfeed.xml"
feed = feedparser.parse(url)

print(feed.feed.title)
for entry in feed.entries[:3]:
    print("-", entry.title)
```

## Bài 2.7 – 2.9 – Chuyển đổi sang JSON

```python
import json

data = {
    "company": "ABC",
    "employees": [
        {"name": "An", "unit": "IT"},
        {"name": "Binh", "unit": "HR"},
        {"name": "Chi", "unit": "IT"}
    ]
}

print(json.dumps(data, indent=4, ensure_ascii=False))
```

## Bài 2.10 – Thống kê nhân viên theo đơn vị

```python
from collections import Counter

employees = [
    {"name": "An", "unit": "IT"},
    {"name": "Binh", "unit": "HR"},
    {"name": "Chi", "unit": "IT"},
    {"name": "Dung", "unit": "Finance"}
]

counter = Counter(e["unit"] for e in employees)

for unit, count in counter.items():
    print(unit, ":", count)
```
