1. Khái niệm của XML:
- XML là ngôn ngữ đánh dấu mở rộng được thiết kế để lưu trữ và trao đổi dữ liệu một cách linh hoạt giữa các hệ thống khác nhau
- XML không phải là ngôn ngữ lập trình mà là ngôn ngữ mô tả dữ liệu
- Dữ liệu trong XML được tổ chức theo dạng cây, với các thẻ do người dùng tự định nghĩa để mô tả nội dung
- XML phân tách rõ ràng giữa dữ liệu và cách hiện thị 
- XML được sử dụng phổ biến để trao đổi dữ liệu giữa các ứng dụng web, hệ thống hoặc nền tảng khác nhau, nhờ tính độc lập và dễ đọc

2. Khái niệm về DOM
- DOM là mô hình đối tượng tài liệu, dùng để biểu diễn cấu trúc của tài liệu XML hoặc HTML dưới dạng cây
- DOM xem mỗi thành phần trong tài liệu( thẻ, thuộc tính, nội dung văn bản,...) là một đối tượng
- Các đối tượng này liên kết với nhau theo quan hệ cha - con, tạo thành cấu trúc cây
- DOM cho phép chương trình(Java, Python...) truy cập, thay đổi, thêm hoặc xóa nội dung của tài liệu 1 cách linh hoạt

3. Cú pháp tài liệu XML:
<?xml version= "1.0" encoding="UFT -8">

4. Khái niệm JSON:
JSON là định dạng trao đổi dữ liệu nhẹ, được thiết kế để lưu trữ và truyền dữ liệu giữa các hệ thống 1 cách đơn giản, dễ đọc và dễ xử lí

5. Hạn chế của việc triển khai JSON trong python
- Không hỗ trợ nhiều kiểu dữ liệu đặc trưng của python
- Mất kiểu dữ liệu gốc khi chuyển đổi
- KHông có chú thích
- Không biểu diễn cấu trúc tham chiếu phức tạp
- Không hỗ trợ dữ liệu nhị phân trực tiếp
- Hiệu năng chưa cao khi xử lí dữ liệu

6. API XML:
- Là giao diện lập trình ứng dụng cho phép chương trình hoặc ứng dụng truy cập, đọc, ghi và xử lí dữ liệu trong tài liệu XML

7. API JSON:
- Là giao diện lập trình ứng dụng cho phép trao đổi dữ liệu giữa các hệ thống thông qua định dạng JSON
