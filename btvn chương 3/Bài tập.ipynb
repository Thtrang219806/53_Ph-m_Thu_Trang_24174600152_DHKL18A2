## Bài 3.2 – Thao tác array cơ bản

```python
import numpy as np

arr = np.arange(10)
print("arr:", arr)
print("dtype:", arr.dtype, "shape:", arr.shape)

arr_odd = arr[arr % 2 == 1]
arr_even = arr[arr % 2 == 0]
print("arr_odd:", arr_odd)
print("arr_even:", arr_even)

arr_update_1 = arr.copy()
arr_update_1[arr_update_1 % 2 == 1] = 100
print("arr_update_1:", arr_update_1)
```

## Bài 3.4 – Thao tác mảng nâng cao

```python
import numpy as np

# 1
arr_zero = np.zeros(10, dtype=int)
arr_zero[5] = 1
print("arr_zero:", arr_zero)

# 2
arr_h = np.arange(10, 21)
print("arr_h reversed:", arr_h[::-1])

# 3
arr_k = np.array([1,2,0,8,2,0,1,3,0,5,0])
print("arr_k non-zero:", arr_k[arr_k != 0])

# 4
arr_add = np.arange(5)
arr_add = np.append(arr_add, [10, 20])
print("arr_add:", arr_add)

# 5
arr_ins = np.arange(10)
arr_ins = np.insert(arr_ins, 5, 100)
print("arr_ins:", arr_ins)

# 6
arr_del = np.delete(arr_ins, [0,1,2])
print("arr_del:", arr_del)
```

## Bài 3.5 – MLB Heights & Weights (giả lập)

```python
import numpy as np

# Giả lập dữ liệu nhỏ
heights = np.array([180, 175, 190, 185])  # cm
weights = np.array([75, 70, 90, 85])      # kg

height_m = heights / 100
bmi = weights / (height_m ** 2)

print("BMI:", bmi)
print("BMI < 21:", bmi[bmi < 21])
print("Avg height:", heights.mean(), "Avg weight:", weights.mean())
print("Max height:", heights.max(), "Max weight:", weights.max())
print("Min height:", heights.min(), "Min weight:", weights.min())
```

## Bài 3.6 – Ma trận 2D

```python
import numpy as np

arr = np.full((3,3), True)
print(arr)

arr1 = np.arange(1, 9).reshape(4,2)
print(arr1[:, ::-1])

print(arr1[::-1, :])
print(arr1[:, ::-1])

arr_nan = np.array([[1,2,3],[np.nan,5,6],[7,np.nan,9]])
print("Has NaN:", np.isnan(arr_nan).any())
arr_nan[np.isnan(arr_nan)] = 0
print(arr_nan)
```

## Bài 3.8 – Trung bình chiều cao theo vị trí (giả lập)

```python
import numpy as np

positions = np.array(['GK','DF','MF','FW','DF'])
heights = np.array([190,180,175,178,182])

for pos in np.unique(positions):
    print(pos, heights[positions == pos].mean())
```
