## Bài 1: Lớp Hình chữ nhật

```python

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def perimeter(self):
        return 2 * (self.length + self.width)

    def area(self):
        return self.length * self.width

    def display(self):
        print("Chiều dài:", self.length)
        print("Chiều rộng:", self.width)
        print("Chu vi:", self.perimeter())
        print("Diện tích:", self.area())

r = Rectangle(5, 3)
r.display()
```

## Bài 2: Lớp Thí Sinh

```python
class TS:
    def __init__(self):
        self.hoten = ""
        self.toan = 0
        self.ly = 0
        self.hoa = 0
    def nhap(self):
        self.hoten = input("Nhập họ tên: ")
        self.toan = float(input("Điểm Toán: "))
        self.ly = float(input("Điểm Lý: "))
        self.hoa = float(input("Điểm Hóa: "))
    def tong(self):
        return self.toan + self.ly + self.hoa

    def xuat(self):
        print(self.hoten, "- Tổng điểm:", self.tong())
ds = []
n = int(input("Nhập số thí sinh: "))
for i in range(n):
    print("\n=== Thí sinh", i+1, "===")
    ts = TS()
    ts.nhap()
    ds.append(ts)
ds.sort(key=lambda x: x.tong(), reverse=True)
print("\n===== DANH SÁCH TRÚNG TUYỂN (điểm chuẩn = 20) =====")
for ts in ds:
    if ts.tong() >= 20:
        ts.xuat()
```

## Bài 3: Lớp Phân số

```python
class PhanSo:
    def __init__(self, tu=0, mau=1):
        self.tu = int(tu)
        self.mau = int(mau)
        if not self.hop_le():
            raise ValueError("Mẫu số không được bằng 0")
    def hop_le(self):
        return self.mau != 0
    @staticmethod
    def uoc_chung_lon_nhat(a, b):
        a, b = abs(a), abs(b)
        while b != 0:
            a, b = b, a % b
        return a
    def rut_gon(self):
        if self.mau == 0:
            raise ValueError("Không thể rút gọn khi mẫu = 0")
        if self.tu == 0:
            self.mau = 1
            return
        g = PhanSo.uoc_chung_lon_nhat(self.tu, self.mau)
        self.tu //= g
        self.mau //= g
        # đưa dấu về tử số
        if self.mau < 0:
            self.tu = -self.tu
            self.mau = -self.mau
    def nhap_tu_ban_phim(self):
        while True:
            s = input("Nhập phân số (ví dụ 3/4 hoặc '3 4'): ").strip()
            try:
                if '/' in s:
                    parts = s.split('/')
                    if len(parts) != 2:
                        raise ValueError
                    tu = int(parts[0].strip())
                    mau = int(parts[1].strip())
                else:
                    parts = s.split()
                    if len(parts) != 2:
                        raise ValueError
                    tu = int(parts[0])
                    mau = int(parts[1])
                if mau == 0:
                    print("Mẫu số không được bằng 0. Nhập lại.")
                    continue
                self.tu = tu
                self.mau = mau
                self.rut_gon()
                break
            except ValueError:
                print("Dữ liệu không hợp lệ. Hãy nhập lại theo dạng 3/4 hoặc '3 4' (các số nguyên).")
    def __str__(self):
        if self.mau == 1:
            return f"{self.tu}"
        return f"{self.tu}/{self.mau}"
    # Ví dụ thêm: chuyển sang số thực
    def to_float(self):
        return self.tu / self.mau
```

## Bài 4-6: Lớp Stack

```python

class Stack:
    def __init__(self, size):
        self.size = size
        self.data = []

    def push(self, x):
        if not self.isFull():
            self.data.append(x)

    def pop(self):
        if not self.isEmpty():
            return self.data.pop()

    def isEmpty(self):
        return len(self.data) == 0

    def isFull(self):
        return len(self.data) == self.size

    def count(self):
        return len(self.data)

    def print_stack(self):
        print(self.data)

s = Stack(5)
s.push(1)
s.push(2)
s.push(3)
s.print_stack()
print("Số phần tử:", s.count())
```

## Bài 7: Lớp Date

```python

class Date:
    def __init__(self, d=1, m=1, y=2000):
        self.d = d
        self.m = m
        self.y = y

    def display(self):
        print(f"{self.d}/{self.m}/{self.y}")

    def next_day(self):
        self.d += 1

dt = Date(30, 9, 2025)
dt.display()
dt.next_day()
dt.display()
```

## Bài 8: Lớp Employee

```python

class Employee:
    def __init__(self, name, birth, start):
        self.name = name
        self.birth = birth
        self.start = start

    def display(self):
        print("Tên:", self.name)
        print("Ngày sinh:", end=" ")
        self.birth.display()
        print("Ngày vào công ty:", end=" ")
        self.start.display()

e = Employee("Trang", Date(1,1,2005), Date(1,9,2024))
e.display()
```

## Bài 9-11: Kế thừa hình học (minh họa)

```python

class HinhChuNhat:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def dien_tich(self):
        return self.a * self.b

class HinhVuong(HinhChuNhat):
    def __init__(self, a):
        super().__init__(a, a)

hv = HinhVuong(4)
print("Diện tích hình vuông:", hv.dien_tich())
```
